## 1. Import the Required Libraries

```python
import tensorflow as tf 
import matplotlib.pyplot as plt 
import pandas as pd
import seaborn as sns 
from tensorflow.keras.utils import image_dataset_from_directory
from tensorflow.keras.layers import Dense, Conv2D, MaxPool2D, Flatten
from tensorflow.keras.models import Sequential
```

## 2. Data Preprocessing

### 2.1 Training Images Preprocessing 

```python
# Path to training dataset directory
train_dir = r"New-Cattle-Diseasee-3/train"
```

```python
# Load the dataset
training_set = image_dataset_from_directory(
    directory=train_dir,
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=True,
    seed=42,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False
)
```

# 2.2 Validation Image Preprocessing

```python
valid_dir = r"New-Cattle-Diseasee-3/valid"
```

```python
validation_set = image_dataset_from_directory(
    directory=valid_dir,
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),
    shuffle=True,
    seed=42,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False
)
```

## 3. Building the Model 

```python
model = Sequential()
```

### 3.1 Building the Convolution Layer 

```python
model.add(Conv2D(filters=32,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=32,kernel_size=3,padding='same',activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
model.add(Conv2D(filters=64,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=64,kernel_size=3,padding='same',activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
model.add(Conv2D(filters=128,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=128,kernel_size=3,padding='same',activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
model.add(Conv2D(filters=256,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=256,kernel_size=3,padding='same',activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
model.add(Flatten())
```

```python
#hidden layer
model.add(Dense(units=1024,activation='relu'))
```

```python
# output layer 
model.add(Dense(units=8,activation='softmax'))
```

```python
# Compile model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model.summary()
```

### 3.2 Training the Model 

```python
training_history = model.fit(x=training_set,
                             validation_data=validation_set,
                             epochs=10)
```
